```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("diabetes.csv")
df.head()
```

```python
# size and column information
print(df.shape)
df.info

# how many values are missing from each column
print(df.isnull().sum())
```

**Histogram of Progression**

```python
plt.hist(df['Progression'], bins=15)

plt.title("Distribution of Disease Progression")
plt.xlabel("Disease Progression")
plt.ylabel("Frequency")

plt.show()
```

**Histogram of BMI**

```python
plt.hist(df['bmi'], bins=15)

plt.title("Distribution of BMI")
plt.xlabel("BMI")
plt.ylabel("Frequency")

plt.show()
```

**BMI boxplot**

```python
sns.boxplot(y=df["bmi"])

plt.title("Boxplot of BMI")
plt.ylabel("Standardised BMI")

plt.show()
```

**Progression boxplot**

```python
sns.boxplot(y=df["Progression"])

plt.title("Boxplot of Progression")
plt.ylabel("Progression")

plt.show()
```

**Correlation Heatmap**

```python
# calculate the correlation matrix
correlation = df.corr(numeric_only=True)

# create heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlation,
            annot=True,
            cmap="coolwarm",
            fmt=".2f")

plt.title("Correlation Heatmap of the Diabetes Dataset")
plt.show()
```

**BMI vs Progression scatter plot**

```python
plt.scatter(df["bmi"], df["Progression"], alpha=0.7)

plt.title("BMI vs Disease Progression")
plt.xlabel("Standardised BMI")
plt.ylabel("Disease Progression")

plt.show()
```

**s5 vs Progression**

```python
plt.scatter(df["s5"], df["Progression"], alpha=0.7)

plt.title("Serum Triglycerides (s5) vs Disease Progression")
plt.xlabel("Standardised s5")
plt.ylabel("Disease Progression")

plt.show()
```

**Blood pressure vs Progression**

```python
plt.scatter(df["bp"], df["Progression"], alpha=0.7)

plt.title("Blood Pressure vs Disease Progression")
plt.xlabel("Standardised Blood Pressure")
plt.ylabel("Disease Progression")

plt.show()
```
